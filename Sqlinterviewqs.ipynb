````python
## 30 SQL Coding Interview Questions for Freshers (Easy to Hard)

This document provides 30 SQL interview questions designed for freshers, covering various fundamental to intermediate SQL concepts. Each question includes a sample table, the problem statement, the SQL query, and the expected output, along with explanations.

---

### **Table Schema for all questions (if not specified otherwise):**

We'll use a few common tables for consistency. Assume the following tables are available unless a question explicitly defines new ones.

**`Employees` Table:**

```sql
CREATE TABLE Employees (
    employee_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    department_id INT,
    salary DECIMAL(10, 2),
    hire_date DATE
);

INSERT INTO Employees (employee_id, first_name, last_name, department_id, salary, hire_date) VALUES
(1, 'Alice', 'Smith', 101, 60000.00, '2020-01-15'),
(2, 'Bob', 'Johnson', 102, 75000.00, '2019-03-20'),
(3, 'Charlie', 'Brown', 101, 62000.00, '2021-06-01'),
(4, 'Diana', 'Prince', 103, 80000.00, '2018-11-10'),
(5, 'Eve', 'Davis', 102, 70000.00, '2022-02-28'),
(6, 'Frank', 'White', 101, 65000.00, '2020-09-01'),
(7, 'Grace', 'Miller', 103, 82000.00, '2019-07-04'),
(8, 'Heidi', 'Clark', 104, 55000.00, '2023-01-01'),
(9, 'Ivan', 'Garcia', 101, 63000.00, '2021-04-10'),
(10, 'Judy', 'Lopez', 102, 78000.00, '2018-05-15');
```

**`Departments` Table:**

```sql
CREATE TABLE Departments (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(50),
    location VARCHAR(50)
);

INSERT INTO Departments (department_id, department_name, location) VALUES
(101, 'HR', 'New York'),
(102, 'IT', 'San Francisco'),
(103, 'Sales', 'Los Angeles'),
(104, 'Marketing', 'New York');
```

**`Projects` Table:**

```sql
CREATE TABLE Projects (
    project_id INT PRIMARY KEY,
    project_name VARCHAR(100),
    department_id INT,
    start_date DATE,
    end_date DATE
);

INSERT INTO Projects (project_id, project_name, department_id, start_date, end_date) VALUES
(1, 'HR Portal Upgrade', 101, '2023-01-01', '2023-06-30'),
(2, 'New System Rollout', 102, '2022-10-01', '2023-12-31'),
(3, 'Sales Campaign Q1', 103, '2023-01-01', '2023-03-31'),
(4, 'Website Redesign', 104, '2023-04-01', '2023-09-30'),
(5, 'Employee Training', 101, '2023-07-01', '2023-08-31'),
(6, 'Data Migration', 102, '2023-03-01', '2024-02-28');
```

---

### **EASY LEVEL**

#### **Question 1: Select All Employees**

**Problem:** Retrieve all columns for every employee in the `Employees` table.

**SQL Query:**
```sql
SELECT * FROM Employees;
```

**Expected Output:**

| employee_id | first_name | last_name | department_id | salary   | hire_date  |
|-------------|------------|-----------|---------------|----------|------------|
| 1           | Alice      | Smith     | 101           | 60000.00 | 2020-01-15 |
| 2           | Bob        | Johnson   | 102           | 75000.00 | 2019-03-20 |
| 3           | Charlie    | Brown     | 101           | 62000.00 | 2021-06-01 |
| 4           | Diana      | Prince    | 103           | 80000.00 | 2018-11-10 |
| 5           | Eve        | Davis     | 102           | 70000.00 | 2022-02-28 |
| 6           | Frank      | White     | 101           | 65000.00 | 2020-09-01 |
| 7           | Grace      | Miller    | 103           | 82000.00 | 2019-07-04 |
| 8           | Heidi      | Clark     | 104           | 55000.00 | 2023-01-01 |
| 9           | Ivan       | Garcia    | 101           | 63000.00 | 2021-04-10 |
| 10          | Judy       | Lopez     | 102           | 78000.00 | 2018-05-15 |

**Explanation:** The `SELECT *` statement is used to retrieve all columns, and `FROM Employees` specifies the table to query.

---

#### **Question 2: Select Employee Names and Salaries**

**Problem:** Get the first name, last name, and salary of all employees.

**SQL Query:**
```sql
SELECT first_name, last_name, salary FROM Employees;
```

**Expected Output:**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Alice      | Smith     | 60000.00 |
| Bob        | Johnson   | 75000.00 |
| Charlie    | Brown     | 62000.00 |
| Diana      | Prince    | 80000.00 |
| Eve        | Davis     | 70000.00 |
| Frank      | White     | 65000.00 |
| Grace      | Miller    | 82000.00 |
| Heidi      | Clark     | 55000.00 |
| Ivan       | Garcia    | 63000.00 |
| Judy       | Lopez     | 78000.00 |

**Explanation:** We explicitly list the column names (`first_name`, `last_name`, `salary`) we want to retrieve.

---

#### **Question 3: Employees with Salary Greater Than 70000**

**Problem:** Find the first name, last name, and salary of employees who earn more than $70,000.

**SQL Query:**
```sql
SELECT first_name, last_name, salary FROM Employees WHERE salary > 70000;
```

**Expected Output:**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Bob        | Johnson   | 75000.00 |
| Diana      | Prince    | 80000.00 |
| Grace      | Miller    | 82000.00 |
| Judy       | Lopez     | 78000.00 |

**Explanation:** The `WHERE` clause is used to filter records based on the condition `salary > 70000`.

---

#### **Question 4: Employees in Department 101**

**Problem:** List all employees working in department ID 101.

**SQL Query:**
```sql
SELECT * FROM Employees WHERE department_id = 101;
```

**Expected Output:**

| employee_id | first_name | last_name | department_id | salary   | hire_date  |
|-------------|------------|-----------|---------------|----------|------------|
| 1           | Alice      | Smith     | 101           | 60000.00 | 2020-01-15 |
| 3           | Charlie    | Brown     | 101           | 62000.00 | 2021-06-01 |
| 6           | Frank      | White     | 101           | 65000.00 | 2020-09-01 |
| 9           | Ivan       | Garcia    | 101           | 63000.00 | 2021-04-10 |

**Explanation:** Filters employees where `department_id` matches '101'.

---

#### **Question 5: Employees Ordered by Salary**

**Problem:** Get the first name, last name, and salary of all employees, ordered by salary in descending order.

**SQL Query:**
```sql
SELECT first_name, last_name, salary FROM Employees ORDER BY salary DESC;
```

**Expected Output:**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Grace      | Miller    | 82000.00 |
| Diana      | Prince    | 80000.00 |
| Judy       | Lopez     | 78000.00 |
| Bob        | Johnson   | 75000.00 |
| Eve        | Davis     | 70000.00 |
| Frank      | White     | 65000.00 |
| Ivan       | Garcia    | 63000.00 |
| Charlie    | Brown     | 62000.00 |
| Alice      | Smith     | 60000.00 |
| Heidi      | Clark     | 55000.00 |

**Explanation:** The `ORDER BY salary DESC` clause sorts the results by the `salary` column in descending order.

---

#### **Question 6: Employees Hired After 2020**

**Problem:** Find employees hired after January 1, 2020.

**SQL Query:**
```sql
SELECT first_name, last_name, hire_date FROM Employees WHERE hire_date > '2020-01-01';
```

**Expected Output:**

| first_name | last_name | hire_date  |
|------------|-----------|------------|
| Alice      | Smith     | 2020-01-15 |
| Charlie    | Brown     | 2021-06-01 |
| Eve        | Davis     | 2022-02-28 |
| Frank      | White     | 2020-09-01 |
| Ivan       | Garcia    | 2021-04-10 |
| Heidi      | Clark     | 2023-01-01 |

**Explanation:** Filters records where `hire_date` is later than '2020-01-01'.

---

#### **Question 7: Count of Employees**

**Problem:** Get the total number of employees.

**SQL Query:**
```sql
SELECT COUNT(employee_id) AS total_employees FROM Employees;
```

**Expected Output:**

| total_employees |
|-----------------|
| 10              |

**Explanation:** The `COUNT()` aggregate function counts the number of non-NULL values in the `employee_id` column. `AS total_employees` gives an alias to the result column.

---

#### **Question 8: Average Salary**

**Problem:** Calculate the average salary of all employees.

**SQL Query:**
```sql
SELECT AVG(salary) AS average_salary FROM Employees;
```

**Expected Output:**

| average_salary |
|----------------|
| 69000.00       |

**Explanation:** The `AVG()` aggregate function calculates the average value of the `salary` column.

---

#### **Question 9: Employees with Salary Between 60000 and 70000**

**Problem:** Find employees whose salary is between $60,000 and $70,000 (inclusive).

**SQL Query:**
```sql
SELECT first_name, last_name, salary FROM Employees WHERE salary BETWEEN 60000 AND 70000;
```

**Expected Output:**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Alice      | Smith     | 60000.00 |
| Charlie    | Brown     | 62000.00 |
| Eve        | Davis     | 70000.00 |
| Frank      | White     | 65000.00 |
| Ivan       | Garcia    | 63000.00 |

**Explanation:** The `BETWEEN` operator filters values within a specified range, inclusive of the start and end values.

---

#### **Question 10: Employees in HR or IT Departments**

**Problem:** Get employees from either the 'HR' (department_id 101) or 'IT' (department_id 102) departments.

**SQL Query:**
```sql
SELECT first_name, last_name, department_id FROM Employees WHERE department_id IN (101, 102);
```

**Expected Output:**

| first_name | last_name | department_id |
|------------|-----------|---------------|
| Alice      | Smith     | 101           |
| Bob        | Johnson   | 102           |
| Charlie    | Brown     | 101           |
| Eve        | Davis     | 102           |
| Frank      | White     | 101           |
| Ivan       | Garcia    | 101           |
| Judy       | Lopez     | 102           |

**Explanation:** The `IN` operator allows specifying multiple values in a `WHERE` clause.

---

### **MEDIUM LEVEL**

#### **Question 11: Employee and Department Names (INNER JOIN)**

**Problem:** List all employees along with their department names.

**SQL Query:**
```sql
SELECT E.first_name, E.last_name, D.department_name
FROM Employees AS E
INNER JOIN Departments AS D ON E.department_id = D.department_id;
```

**Expected Output:**

| first_name | last_name | department_name |
|------------|-----------|-----------------|
| Alice      | Smith     | HR              |
| Bob        | Johnson   | IT              |
| Charlie    | Brown     | HR              |
| Diana      | Prince    | Sales           |
| Eve        | Davis     | IT              |
| Frank      | White     | HR              |
| Grace      | Miller    | Sales           |
| Heidi      | Clark     | Marketing       |
| Ivan       | Garcia    | HR              |
| Judy       | Lopez     | IT              |

**Explanation:** An `INNER JOIN` combines rows from two tables (`Employees` and `Departments`) when there's a match in the join condition (`E.department_id = D.department_id`). Aliases (`AS E`, `AS D`) are used for brevity.

---

#### **Question 12: Departments with No Employees (LEFT JOIN)**

**Problem:** Find all departments and list employees if they exist. Show departments even if they have no employees.

**SQL Query:**
```sql
SELECT D.department_name, E.first_name, E.last_name
FROM Departments AS D
LEFT JOIN Employees AS E ON D.department_id = E.department_id;
```

**Expected Output:**

| department_name | first_name | last_name |
|-----------------|------------|-----------|
| HR              | Alice      | Smith     |
| HR              | Charlie    | Brown     |
| HR              | Frank      | White     |
| HR              | Ivan       | Garcia    |
| IT              | Bob        | Johnson   |
| IT              | Eve        | Davis     |
| IT              | Judy       | Lopez     |
| Sales           | Diana      | Prince    |
| Sales           | Grace      | Miller    |
| Marketing       | Heidi      | Clark     |

**Explanation:** A `LEFT JOIN` returns all rows from the left table (`Departments`) and the matching rows from the right table (`Employees`). If there's no match, NULLs are returned for the right table's columns.

---

#### **Question 13: Total Salary per Department (GROUP BY)**

**Problem:** Calculate the total salary expenditure for each department.

**SQL Query:**
```sql
SELECT D.department_name, SUM(E.salary) AS total_salary_expenditure
FROM Employees AS E
JOIN Departments AS D ON E.department_id = D.department_id
GROUP BY D.department_name;
```

**Expected Output:**

| department_name | total_salary_expenditure |
|-----------------|--------------------------|
| HR              | 250000.00                |
| IT              | 223000.00                |
| Marketing       | 55000.00                 |
| Sales           | 162000.00                |

**Explanation:** `GROUP BY D.department_name` groups the employees by their department, and `SUM(E.salary)` calculates the sum of salaries for each group.

---

#### **Question 14: Departments with More Than 3 Employees (HAVING)**

**Problem:** Find the department names that have more than 3 employees.

**SQL Query:**
```sql
SELECT D.department_name, COUNT(E.employee_id) AS num_employees
FROM Employees AS E
JOIN Departments AS D ON E.department_id = D.department_id
GROUP BY D.department_name
HAVING COUNT(E.employee_id) > 3;
```

**Expected Output:**

| department_name | num_employees |
|-----------------|---------------|
| HR              | 4             |

**Explanation:** `HAVING COUNT(E.employee_id) > 3` filters the results of the `GROUP BY` clause, showing only those departments where the count of employees exceeds 3. `WHERE` filters rows *before* aggregation, while `HAVING` filters groups *after* aggregation.

---

#### **Question 15: Employees Earning More Than Department Average (Subquery)**

**Problem:** List employees who earn more than the average salary of their respective departments.

**SQL Query:**
```sql
SELECT E.first_name, E.last_name, E.salary, D.department_name
FROM Employees AS E
JOIN Departments AS D ON E.department_id = D.department_id
WHERE E.salary > (
    SELECT AVG(salary)
    FROM Employees
    WHERE department_id = E.department_id
);
```

**Expected Output:**

| first_name | last_name | salary   | department_name |
|------------|-----------|----------|-----------------|
| Bob        | Johnson   | 75000.00 | IT              |
| Frank      | White     | 65000.00 | HR              |
| Judy       | Lopez     | 78000.00 | IT              |

**Explanation:** The subquery calculates the average salary for the current employee's department. The outer query then compares each employee's salary to this departmental average.

---

#### **Question 16: Second Highest Salary in Each Department (Window Function)**

**Problem:** Find the employee(s) with the second-highest salary in each department.

**SQL Query:**
```sql
WITH RankedSalaries AS (
    SELECT
        employee_id,
        first_name,
        last_name,
        department_id,
        salary,
        DENSE_RANK() OVER (PARTITION BY department_id ORDER BY salary DESC) as salary_rank
    FROM Employees
)
SELECT
    RS.first_name, RS.last_name, RS.salary, D.department_name
FROM RankedSalaries AS RS
JOIN Departments AS D ON RS.department_id = D.department_id
WHERE RS.salary_rank = 2;
```

**Expected Output:**

| first_name | last_name | salary   | department_name |
|------------|-----------|----------|-----------------|
| Charlie    | Brown     | 62000.00 | HR              |
| Eve        | Davis     | 70000.00 | IT              |
| Diana      | Prince    | 80000.00 | Sales           |

**Explanation:** A Common Table Expression (CTE) `RankedSalaries` is used to rank employees within each department (`PARTITION BY department_id`) by their salary in descending order using `DENSE_RANK()`. The outer query then selects employees with a `salary_rank` of 2.

---

#### **Question 17: Employees and Projects (LEFT JOIN)**

**Problem:** List all departments and any projects associated with them. Include departments with no projects.

**SQL Query:**
```sql
SELECT D.department_name, P.project_name
FROM Departments AS D
LEFT JOIN Projects AS P ON D.department_id = P.department_id;
```

**Expected Output:**

| department_name | project_name      |
|-----------------|-------------------|
| HR              | HR Portal Upgrade |
| HR              | Employee Training |
| IT              | New System Rollout|
| IT              | Data Migration    |
| Sales           | Sales Campaign Q1 |
| Marketing       | Website Redesign  |

**Explanation:** A `LEFT JOIN` ensures all departments are listed, and projects are included where a `department_id` match exists.

---

#### **Question 18: Employees with 'o' in their Name**

**Problem:** Find employees whose first name or last name contains the letter 'o'.

**SQL Query:**
```sql
SELECT first_name, last_name
FROM Employees
WHERE first_name LIKE '%o%' OR last_name LIKE '%o%';
```

**Expected Output:**

| first_name | last_name |
|------------|-----------|
| Bob        | Johnson   |
| Charlie    | Brown     |
| Judy       | Lopez     |

**Explanation:** The `LIKE` operator combined with the wildcard character `%` matches patterns. `%o%` means any string containing 'o'.

---

#### **Question 19: Projects Starting in 2023**

**Problem:** List all projects that started in the year 2023.

**SQL Query:**
```sql
SELECT project_name, start_date
FROM Projects
WHERE YEAR(start_date) = 2023;
-- Or for SQL Server/MySQL: WHERE start_date BETWEEN '2023-01-01' AND '2023-12-31';
-- Or for PostgreSQL: WHERE EXTRACT(YEAR FROM start_date) = 2023;
```

**Expected Output:**

| project_name      | start_date |
|-------------------|------------|
| HR Portal Upgrade | 2023-01-01 |
| Sales Campaign Q1 | 2023-01-01 |
| Website Redesign  | 2023-04-01 |
| Employee Training | 2023-07-01 |
| Data Migration    | 2023-03-01 |

**Explanation:** The `YEAR()` function (or equivalent for your SQL dialect) extracts the year from a date column, allowing for filtering by year.

---

#### **Question 20: Maximum and Minimum Salary in the Company**

**Problem:** Find the highest and lowest salary among all employees.

**SQL Query:**
```sql
SELECT MAX(salary) AS max_salary, MIN(salary) AS min_salary FROM Employees;
```

**Expected Output:**

| max_salary | min_salary |
|------------|------------|
| 82000.00   | 55000.00   |

**Explanation:** `MAX()` and `MIN()` are aggregate functions used to find the maximum and minimum values in a column, respectively.

---

### **HARD LEVEL**

#### **Question 21: Employees with Salary Higher than All Employees in Department 104 (Subquery with ALL)**

**Problem:** Find employees whose salary is higher than the salary of *all* employees in department 104.

**SQL Query:**
```sql
SELECT first_name, last_name, salary
FROM Employees
WHERE salary > ALL (
    SELECT salary
    FROM Employees
    WHERE department_id = 104
);
```

**Expected Output:**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Alice      | Smith     | 60000.00 |
| Bob        | Johnson   | 75000.00 |
| Charlie    | Brown     | 62000.00 |
| Diana      | Prince    | 80000.00 |
| Eve        | Davis     | 70000.00 |
| Frank      | White     | 65000.00 |
| Grace      | Miller    | 82000.00 |
| Ivan       | Garcia    | 63000.00 |
| Judy       | Lopez     | 78000.00 |

**Explanation:** The subquery returns a set of salaries for department 104. The `ALL` operator ensures that the employee's salary must be greater than *every* salary in that set.

---

#### **Question 22: Cumulative Salary by Department and Hire Date (Window Function)**

**Problem:** Calculate the cumulative salary sum for employees within each department, ordered by hire date.

**SQL Query:**
```sql
SELECT
    employee_id,
    first_name,
    last_name,
    department_id,
    hire_date,
    salary,
    SUM(salary) OVER (PARTITION BY department_id ORDER BY hire_date) AS cumulative_department_salary
FROM Employees;
```

**Expected Output:**

| employee_id | first_name | last_name | department_id | hire_date  | salary   | cumulative_department_salary |
|-------------|------------|-----------|---------------|----------|----------|------------------------------|
| 1           | Alice      | Smith     | 101           | 2020-01-15 | 60000.00 | 60000.00                     |
| 6           | Frank      | White     | 101           | 2020-09-01 | 65000.00 | 125000.00                    |
| 3           | Charlie    | Brown     | 101           | 2021-06-01 | 62000.00 | 187000.00                    |
| 9           | Ivan       | Garcia    | 101           | 2021-04-10 | 63000.00 | 250000.00                    |
| 10          | Judy       | Lopez     | 102           | 2018-05-15 | 78000.00 | 78000.00                     |
| 2           | Bob        | Johnson   | 102           | 2019-03-20 | 75000.00 | 153000.00                    |
| 5           | Eve        | Davis     | 102           | 2022-02-28 | 70000.00 | 223000.00                    |
| 4           | Diana      | Prince    | 103           | 2018-11-10 | 80000.00 | 80000.00                     |
| 7           | Grace      | Miller    | 103           | 2019-07-04 | 82000.00 | 162000.00                    |
| 8           | Heidi      | Clark     | 104           | 2023-01-01 | 55000.00 | 55000.00                     |

**Explanation:** `SUM(salary) OVER (PARTITION BY department_id ORDER BY hire_date)` calculates a running total of salaries. `PARTITION BY department_id` ensures the sum resets for each department, and `ORDER BY hire_date` defines the order within each partition.

---

#### **Question 23: Departments with No Projects (Subquery with NOT EXISTS)**

**Problem:** Identify departments that currently have no projects assigned to them.

**SQL Query:**
```sql
SELECT department_name
FROM Departments AS D
WHERE NOT EXISTS (
    SELECT 1
    FROM Projects AS P
    WHERE P.department_id = D.department_id
);
```

**Expected Output:**

| department_name |
|-----------------|

*(Based on the sample data, all departments have projects. If Department 105 existed without projects, it would show here.)*

**Explanation:** The `NOT EXISTS` clause checks if any rows are returned by the subquery. If the subquery finds no projects for a given department ID, then `NOT EXISTS` evaluates to true, and that department's name is returned.

**Let's add a department without projects to see this work:**

```sql
INSERT INTO Departments (department_id, department_name, location) VALUES (105, 'Research', 'Boston');
```

Now, re-running the query:

**Expected Output (with new data):**

| department_name |
|-----------------|
| Research        |

---

#### **Question 24: Employees Who Are the Highest Earner in Their Department (CTE + Subquery)**

**Problem:** Find the name and salary of employees who earn the highest salary in their respective departments.

**SQL Query:**
```sql
WITH DeptMaxSalary AS (
    SELECT department_id, MAX(salary) AS max_dept_salary
    FROM Employees
    GROUP BY department_id
)
SELECT E.first_name, E.last_name, E.salary, D.department_name
FROM Employees AS E
JOIN Departments AS D ON E.department_id = D.department_id
JOIN DeptMaxSalary AS DMS ON E.department_id = DMS.department_id AND E.salary = DMS.max_dept_salary;
```

**Expected Output:**

| first_name | last_name | salary   | department_name |
|------------|-----------|----------|-----------------|
| Frank      | White     | 65000.00 | HR              |
| Judy       | Lopez     | 78000.00 | IT              |
| Grace      | Miller    | 82000.00 | Sales           |
| Heidi      | Clark     | 55000.00 | Marketing       |

**Explanation:** The CTE `DeptMaxSalary` first identifies the maximum salary for each department. Then, the main query joins `Employees` with `Departments` and `DeptMaxSalary` to filter for employees whose salary matches their department's maximum salary.

---

#### **Question 25: Employees Hired Most Recently in Each Department (Window Function)**

**Problem:** Find the employee(s) who were hired most recently in each department.

**SQL Query:**
```sql
WITH RecentHires AS (
    SELECT
        employee_id,
        first_name,
        last_name,
        department_id,
        hire_date,
        RANK() OVER (PARTITION BY department_id ORDER BY hire_date DESC) as hire_rank
    FROM Employees
)
SELECT
    RH.first_name, RH.last_name, RH.hire_date, D.department_name
FROM RecentHires AS RH
JOIN Departments AS D ON RH.department_id = D.department_id
WHERE RH.hire_rank = 1;
```

**Expected Output:**

| first_name | last_name | hire_date  | department_name |
|------------|-----------|------------|-----------------|
| Ivan       | Garcia    | 2021-04-10 | HR              |
| Eve        | Davis     | 2022-02-28 | IT              |
| Grace      | Miller    | 2019-07-04 | Sales           |
| Heidi      | Clark     | 2023-01-01 | Marketing       |

**Explanation:** Similar to Question 16, `RANK() OVER (PARTITION BY department_id ORDER BY hire_date DESC)` assigns a rank based on hire date within each department. We then select those with `hire_rank = 1`.

---

#### **Question 26: Number of Projects per Department Location**

**Problem:** Count the number of projects for each department location.

**SQL Query:**
```sql
SELECT D.location, COUNT(P.project_id) AS num_projects
FROM Departments AS D
JOIN Projects AS P ON D.department_id = P.department_id
GROUP BY D.location
HAVING COUNT(P.project_id) > 0; -- Optional: only show locations with projects
```

**Expected Output:**

| location      | num_projects |
|---------------|--------------|
| New York      | 3            |
| San Francisco | 2            |
| Los Angeles   | 1            |

**Explanation:** Joins `Departments` and `Projects` to link projects to locations, then `GROUP BY D.location` aggregates the count of projects for each distinct location.

---

#### **Question 27: Employee Salary Percentile (Window Function)**

**Problem:** Calculate the percentile rank of each employee's salary within their department.

**SQL Query:**
```sql
SELECT
    employee_id,
    first_name,
    last_name,
    department_id,
    salary,
    NTILE(100) OVER (PARTITION BY department_id ORDER BY salary) AS salary_percentile
    -- Or PERCENT_RANK() OVER (PARTITION BY department_id ORDER BY salary) for a more precise percentile (0 to 1)
FROM Employees;
```

**Expected Output (using NTILE(100)) - example for HR department:**

| employee_id | first_name | last_name | department_id | salary   | salary_percentile |
|-------------|------------|-----------|---------------|----------|-------------------|
| 1           | Alice      | Smith     | 101           | 60000.00 | 25                |
| 3           | Charlie    | Brown     | 101           | 62000.00 | 50                |
| 9           | Ivan       | Garcia    | 101           | 63000.00 | 75                |
| 6           | Frank      | White     | 101           | 65000.00 | 100               |

*(The exact percentile numbers will depend on the distribution within each partition and the `NTILE` value chosen.)*

**Explanation:** `NTILE(100) OVER (PARTITION BY department_id ORDER BY salary)` divides employees within each department into 100 groups based on their salary, assigning a percentile rank. `PERCENT_RANK()` gives a continuous percentile from 0 to 1.

---

#### **Question 28: Find Duplicate Salaries (Self-Join)**

**Problem:** Find employees who have the same salary as another employee (excluding themselves).

**SQL Query:**
```sql
SELECT E1.first_name, E1.last_name, E1.salary
FROM Employees AS E1
JOIN Employees AS E2
    ON E1.salary = E2.salary AND E1.employee_id != E2.employee_id;
```

**Expected Output:**

*(Based on the provided data, there are no exact duplicate salaries. If there were, they would appear here. Let's add a duplicate salary to demonstrate:)*

```sql
INSERT INTO Employees (employee_id, first_name, last_name, department_id, salary, hire_date) VALUES
(11, 'Zara', 'Khan', 101, 60000.00, '2023-05-01');
```

Now, re-running the query:

**Expected Output (with new data):**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Alice      | Smith     | 60000.00 |
| Zara       | Khan      | 60000.00 |

**Explanation:** A `SELF-JOIN` joins a table to itself. Here, `Employees` is joined with itself (`E1` and `E2`) to compare salaries. The condition `E1.employee_id != E2.employee_id` ensures we don't compare an employee with themselves.

---

#### **Question 29: Employees with Salary Below the Company's Overall Average (Subquery)**

**Problem:** List all employees whose salary is below the overall average salary of the company.

**SQL Query:**
```sql
SELECT first_name, last_name, salary
FROM Employees
WHERE salary < (
    SELECT AVG(salary)
    FROM Employees
);
```

**Expected Output:**

| first_name | last_name | salary   |
|------------|-----------|----------|
| Alice      | Smith     | 60000.00 |
| Charlie    | Brown     | 62000.00 |
| Eve        | Davis     | 70000.00 |
| Frank      | White     | 65000.00 |
| Heidi      | Clark     | 55000.00 |
| Ivan       | Garcia    | 63000.00 |

*(Assuming average salary is ~69000.00 based on initial data)*

**Explanation:** A simple subquery calculates the overall average salary, which is then used in the `WHERE` clause of the outer query to filter employees.

---

#### **Question 30: Average Project Duration per Department (CTE)**

**Problem:** Calculate the average duration (in days) of projects for each department.

**SQL Query:**
```sql
WITH ProjectDurations AS (
    SELECT
        department_id,
        project_name,
        DATEDIFF(end_date, start_date) AS duration_days
        -- Or (end_date - start_date) for PostgreSQL/Oracle if result is interval then extract days
        -- Or JULIANDAY(end_date) - JULIANDAY(start_date) for SQLite
    FROM Projects
    WHERE end_date IS NOT NULL AND start_date IS NOT NULL
)
SELECT
    D.department_name,
    AVG(PD.duration_days) AS average_project_duration_days
FROM ProjectDurations AS PD
JOIN Departments AS D ON PD.department_id = D.department_id
GROUP BY D.department_name;
```

**Expected Output:**

| department_name | average_project_duration_days |
|-----------------|-------------------------------|
| HR              | 121.5                         |
| IT              | 335.5                         |
| Sales           | 90                            |
| Marketing       | 183                           |

**Explanation:** The CTE `ProjectDurations` first calculates the duration of each project using `DATEDIFF()` (or equivalent date difference function). The main query then joins this CTE with the `Departments` table and calculates the average duration grouped by department name. `WHERE end_date IS NOT NULL AND start_date IS NOT NULL` handles cases where dates might be missing.
````
